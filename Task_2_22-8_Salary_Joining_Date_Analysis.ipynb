```python
def normalize_datetime(value):
    """
    Convert a date/time value into a consistent UTC-aware timestamp.
    Invalid values return NaT.
    """
    try:
        dt = pd.to_datetime(value, errors="coerce", utc=True)
        return dt
    except Exception:
        return pd.NaT
```

```python
test_dates = [
    "2024-11-15",
    "2024-11-15 10:30:00",
    "2024-11-15 10:30:00+05:30",
    "2024-11-15 05:00:00+00:00",
    "invalid-date"
]

for value in test_dates:
    print(value, "->", normalize_datetime(value))
```

```python
df["estimated_joining_date"] = (
    df["survey_date"] -
    pd.to_timedelta(df["experience_years"] * 365.25, unit="D")
)
```

```python
df[
    [
        "survey_date",
        "experience_years",
        "estimated_joining_date"
    ]
].head(10)
```

```python
df["joining_year"] = df["estimated_joining_date"].dt.year
df["joining_month"] = df["estimated_joining_date"].dt.month
df["joining_day"] = df["estimated_joining_date"].dt.day
df["joining_quarter"] = df["estimated_joining_date"].dt.quarter
df["joining_day_of_week"] = df["estimated_joining_date"].dt.day_name()
```

```python
df[
    [
        "estimated_joining_date",
        "joining_year",
        "joining_month",
        "joining_day",
        "joining_quarter",
        "joining_day_of_week"
    ]
].head(10)
```

```python
today = pd.Timestamp.today().normalize()

df["tenure_years"] = (
    (today - df["estimated_joining_date"]).dt.days / 365.25
)
```

```python
df["tenure_years"] = df["tenure_years"].round(2)
```

```python
df[
    [
        "estimated_joining_date",
        "tenure_years"
    ]
].head(10)
```

```python
df["tenure_months"] = (
    df["tenure_years"] * 12
).round().astype("Int64")
```

```python
df[
    [
        "estimated_joining_date",
        "tenure_years",
        "tenure_months"
    ]
].head(10)
```

```python
print("Minimum tenure:", df["tenure_years"].min())
print("Maximum tenure:", df["tenure_years"].max())

print(
    "Negative tenure records:",
    (df["tenure_years"] < 0).sum()
)
```

```python
df["tenure_category"] = pd.cut(
    df["tenure_years"],
    bins=[-1, 2, 5, 10, 20, np.inf],
    labels=[
        "0-2 Years",
        "3-5 Years",
        "6-10 Years",
        "11-20 Years",
        "20+ Years"
    ]
)
```
